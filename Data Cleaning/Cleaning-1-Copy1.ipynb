### Price
Selecting a common unit

```python
df['Price'].str.split(' ').str.get(1).value_counts()
```

```python
df['Price'].str.split(' ').str.get(0).str[3:].astype('float')
```

```python
def Price_to_Cr(series):
    unit = series.strip().split(' ')[1]
    price = float(series.strip().split(' ')[0][3:])
    if (unit =='Lakh'):
        price = price / 100
    elif (unit =='Arab'):
        price = price * 100

    series = price
    return series
```

```python
df['Price'].apply(Price_to_Cr)
```

```python
df['Price'] = df['Price'].apply(Price_to_Cr)
```

```python
df['Price'].sample(10)
```

```python
df.rename(columns={"Price":'Price(Cr)'},inplace=True)
```

### Area
Converting Kanal to Marla. <br>
1 Kanal = 20 Marla

```python
df['Area'].sample(5)
```

```python
# No other unit for Area
df['Area'].str.split().str.get(1).value_counts()
```

```python
def Areaunitconverter(series):
    unit = series.split()[1].strip()
    no = float(series.split()[0])
    if unit == 'Kanal':
        no *= 20

    return no
```

```python
df['Area'].apply(Areaunitconverter).tail(5)
```

```python
df['Area'].tail(5)
```

```python
df['Area'] = df['Area'].apply(Areaunitconverter)
```

```python
df.rename(columns={'Area':'Area(Marla)'},inplace=True)
```
