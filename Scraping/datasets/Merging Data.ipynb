```python
def mergingDFs():
    import pandas as pd
    import os

    global files
    files = os.listdir("C:/Users/Wajih/Work (DataScience)/CAPSTONE PROJECTS/Real Estate Project/Zameen.com/Scraping/datasets")

    files = [i for i in files if i.startswith('final_data')]

    original_df = pd.DataFrame()

    for fileName in files:
        temp = pd.read_csv(fileName)
    
        original_df = pd.concat((original_df,temp),ignore_index=True)
        original_df.drop_duplicates(inplace=True)

    return original_df
```

```python
mergingDFs()
```
