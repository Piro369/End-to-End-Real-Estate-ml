```python
import numpy as np
import pandas as pd
```

```python
X_train = pd.read_csv('../Feature Selection/X_train_flats_test.csv',index_col=0,)
X_test = pd.read_csv('../Feature Selection/X_test_flats_test.csv',index_col=0)
y_train = pd.read_csv('../Feature Selection/y_train_flats.csv',index_col=0)
y_test = pd.read_csv('../Feature Selection/y_test_flats.csv',index_col=0)
```

```python
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.metrics import r2_score,mean_absolute_error
```

```python
ridge = Ridge(alpha=94)
ridge.fit(X_train,y_train)
y_pred = ridge.predict(X_test)
# y_pred = np.expm1(y_pred)
mean_absolute_error(y_test,y_pred)
```
