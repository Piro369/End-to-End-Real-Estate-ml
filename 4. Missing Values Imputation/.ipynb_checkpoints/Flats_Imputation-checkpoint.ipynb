```python
# First we remove the featuress we dont need

cols_to_remove = [
    'Amenities: Main Features', 'Amenities: Rooms',
    'Amenities: Healthcare Recreational',
    'Amenities: Other Facilities','IsPrimeLoc','Service Elevators in Building',
    'Sauna','Jacuzzi', 'BedBath_Diff','Swimming Pool','Lobby in Building'
]

flats.drop(columns=cols_to_remove,inplace = True)
```

### Bedroom(s) & Bath(s)

*Removing the rows where Bath(s) or  Bedroom(s) are null*

```python
flats.dropna(subset=['Bath(s)','Bedroom(s)'],inplace=True)
```

```python
flats.shape
```

```python
flats.isnull().sum()
```

### Built in Year

1. First of all we are going to convert the Year column to Category.
2. Then fill the missing values based on Building


```python
flats['Built in year'].value_counts()
```

- Future --> $>2025$
- Immediate / Brand New --> $2025$
- Recently Built --> $2020 - 2024$
- Modern Era --> $2010 - 2019$
- Established --> $2000 - 2009$
- Vintage / Legacy-- > $< 2000$

```python
# Converting 

bins = [-float('inf'),1999,2009,2019,2024,2025,float('inf')]
labels = ['Vintage','Established','Modern Era','Recently Built','Brand New (2025)','Future']

flats['Property era'] =  pd.cut(flats['Built in year'],bins=bins,labels=labels)
```

```python
flats['Property era'].value_counts()
```

```python
flats.groupby('Property era')['Price(Cr)'].mean().sort_index()
```

```python
flats.groupby('Property era')['Area(Marla)'].mean().sort_index()
```

*Old properties have greater prices on average, than the new properties. One reason could be that old properties had greater land rather than the New ones.* 

```python
flats[flats['Property era'].isnull()]['Main Location'].unique()
```

```python
location_modes =  flats.groupby('Building')['Property era'].apply(lambda x : x.mode().iloc[0] if not x.mode().empty else np.nan )
```

```python
location_modes
```

```python
flats['Building'].nunique()
```

```python
flats['Building'].map(location_modes)
```

```python
flats['Property era'] =  flats['Property era'].fillna(flats['Building'].map(location_modes))
```

```python
flats['Property era'].isnull().sum()
```

```python
# Filling the value with the mode now

flats['Property era'] = flats['Property era'].fillna(flats['Property era'].mode()[0])
```

```python
# Removing Built in Year
flats.drop(columns='Built in year',inplace=True)
```

```python
flats.isnull().sum()
```

### Parking Spaces, Floor, Elevators, Floor in Building

```python
# Parking Spaces

mode_parking = flats.groupby('Building')['Parking Spaces'].apply(lambda x: x.mode().iloc[0] if not  x.mode().empty else np.nan)

flats['Parking Spaces'] = flats['Parking Spaces'].fillna(flats['Building'].map(mode_parking))
```

```python
flats['Parking Spaces'].isnull().sum()
```

```python
flats['Parking Spaces'] = flats['Parking Spaces'].fillna(flats['Parking Spaces'].mode()[0])
```

```python
# Floor

mode_floor = flats.groupby('Building')['Floor'].apply(lambda x: x.mode().iloc[0] if not  x.mode().empty else np.nan)

flats['Floor'] = flats['Floor'].fillna(flats['Building'].map(mode_floor))
```

```python
flats['Floor'].isnull().sum()
```

```python
flats['Floor'] = flats['Floor'].fillna(flats['Floor'].mode()[0])
```

```python
# Elevators

mode_elevators = flats.groupby('Building')['Elevators'].apply(lambda x: x.mode().iloc[0] if not  x.mode().empty else np.nan)

flats['Elevators'] = flats['Elevators'].fillna(flats['Building'].map(mode_elevators))
```

```python
flats['Elevators'].isnull().sum()
```

```python
flats['Elevators'] = flats['Elevators'].fillna(flats['Elevators'].mode()[0])
```

```python
# Floor

mode_floorBuilding = flats.groupby('Building')['Floor in Building'].apply(lambda x: x.mode().iloc[0] if not  x.mode().empty else np.nan)

flats['Floor in Building'] = flats['Floor in Building'].fillna(flats['Building'].map(mode_floorBuilding))
```

```python
flats['Floor in Building'].isnull().sum()
```

```python
flats['Floor in Building'] = flats['Floor in Building'].fillna(flats['Floor in Building'].mode()[0])
```

```python
flats.isnull().sum()
```

```python
# Changing to categories (Floor, Floor in Building, Elevators)
```

```python
# Floor
```

```python
floor_bins = [0, 1, 5, 10, float('inf')]
floor_labels = ['Ground/First Floor', 'Lower Floors', 'Mid-Level', 'High-Rise']

flats['Floor Level'] = pd.cut(flats['Floor'], bins=floor_bins, labels=floor_labels)
```

```python
flats['Floor Level'].value_counts()
```

```python
# Floor in Building
```

```python
bins = [0,1,4,12,25,float('inf')]
labels = ['Ground/Flat', 'Lower Levels','Mid Rise Core','High Rise','Skyscarper']

flats['Building Type'] =  pd.cut(flats['Floor in Building'],bins=bins,labels=labels)
```

```python
flats['Building Type'].value_counts()
```

```python
# Elevators
```

```python
bins = [-1, 1, 2, 4, float('inf')]
labels = ['Single/None', 'Dual Setup', 'Multiple Bank', 'High-Capacity']

flats['Elevator Capacity'] = pd.cut(flats['Elevators'], bins=bins, labels=labels)
```

```python
flats['Elevator Capacity'].value_counts()
```

### Servant Quarters, Kitchens, Store Rooms

Using Iterative Imputer(Tree based model) to fill the missing values because we haven't scaled our data.

Making a pipeline and doing a Randomized Search CV to find the best hyper parameters for imputing the values.

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer,SimpleImputer,KNNImputer

from sklearn.ensemble import RandomForestRegressor,ExtraTreesRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from xgboost import XGBRegressor

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.model_selection import RandomizedSearchCV,GridSearchCV

from sklearn import set_config

set_config(transform_output='pandas')
```

```python
cols_to_impute = ['Bath(s)', 'Area(Marla)', 'Bedroom(s)','Servant Quarters', 'Kitchens', 'Store Rooms']
```

```python
trf_simple_imputer = ColumnTransformer(
    [
        ('SimpleImputer',SimpleImputer(),cols_to_impute)
    ],
    remainder='passthrough',verbose_feature_names_out=False
)

trf_iter_imputer = ColumnTransformer(
    [
        ('IterImputer',IterativeImputer(estimator=ExtraTreesRegressor(n_estimators=100, max_depth=10, random_state=42)),cols_to_impute)
    ],
    remainder='passthrough',verbose_feature_names_out=False
)

trf_knn_imputer = ColumnTransformer(
    [
        ('KnnImputer',KNNImputer(),cols_to_impute)
    ],
    remainder='passthrough',verbose_feature_names_out=False
)
```

```python
final_pipeline = Pipeline(
    [
        ('Impute',None),
        ('Model',XGBRegressor(n_estimators=100))
    ]
)
```

```python
param_grid = [
    {
        "Impute":[trf_simple_imputer],
        'Impute__SimpleImputer__strategy':['mean','most_frequent']
    },
    {
        "Impute":[trf_iter_imputer],
        "Impute__IterImputer__max_iter":[10,20,30],
        "Impute__IterImputer__n_nearest_features": [3, 5, 10],
        "Impute__IterImputer__imputation_order":['random','ascending','arabic']
    },
    {
        "Impute":[trf_knn_imputer],
        "Impute__KnnImputer__n_neighbors":[10,20,30],
        "Impute__KnnImputer__weights":['distance',"uniform"]
    }
    
]

```

```python
from sklearn.model_selection import train_test_split

X = flats[['Bath(s)', 'Area(Marla)', 'Bedroom(s)','Servant Quarters', 'Kitchens', 'Store Rooms']]
y = flats['Price(Cr)']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42)
```

```python
search_cv = RandomizedSearchCV(final_pipeline,param_grid,n_iter=20,scoring='r2',n_jobs=-1,cv=5,)

search_cv.fit(X_train,y_train)
```

```python
search_cv.best_score_
```

```python
search_cv.best_estimator_
```

```python
results = pd.DataFrame(search_cv.cv_results_)
```

```python
results.sort_values(by=['mean_test_score'],ascending=False)[['mean_test_score', 'std_test_score','split0_test_score',
       'split1_test_score', 'split2_test_score', 'split3_test_score',
       'split4_test_score']]
```

```python
search_cv.best_params_
```

```python
trf_best_imputer = ColumnTransformer(
    [
        ('Iter',SimpleImputer(strategy='most_frequent'),cols_to_impute)
    ],
    remainder='passthrough',verbose_feature_names_out=False,n_jobs=-1
)

best_pipeline =Pipeline(
    [
        ('Impute',trf_best_imputer),
        ('Model',XGBRegressor(n_estimators=100,n_jobs=-1))
    ]
)
```

```python
best_pipeline.fit(X_train,y_train)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error

y_pred = best_pipeline.predict(X_test)
print('R2 Score',r2_score(y_test,y_pred))
print('Mean Absolute Error',mean_absolute_error(y_test,y_pred))
```
